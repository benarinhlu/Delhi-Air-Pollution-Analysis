```python
# Identify sustained pollution episodes (7-day average > 300)

daily_aqi = (
    ncr_data.groupby(['date', 'city'])['aqi'].mean()
    .groupby(level='date').mean()
    .rename('AQI').to_frame()
    .sort_index()
)

# Calendar-based rolling average
daily_aqi['7_Day_Avg'] = daily_aqi['AQI'].rolling(
    '7D', min_periods=5
).mean()

# Identify consecutive dates above threshold
mask = daily_aqi['7_Day_Avg'] > 300

groups = (
    (~mask | ~mask.shift(fill_value=False) |
     daily_aqi.index.to_series().diff().dt.days.ne(1))
    .cumsum()
)

episodes = (
    daily_aqi[mask]
    .groupby(groups[mask])
    .agg(
        Start=('7_Day_Avg', lambda x: x.index.min()),
        End=('7_Day_Avg', lambda x: x.index.max()),
        Duration=('AQI', 'size'),
        Peak_AQI=('AQI', 'max'),
        Peak_7Day_Avg=('7_Day_Avg', 'max')
    )
)

# Retain episodes overlapping October and November
episodes = episodes[
    episodes['Start'].dt.month.isin([10, 11]) |
    episodes['End'].dt.month.isin([10, 11]) |
    ((episodes['Start'].dt.year == episodes['End'].dt.year) &
     (episodes['Start'].dt.month < 10) &
     (episodes['End'].dt.month > 11))
]

result = episodes.sort_values(
    'Peak_7Day_Avg',
    ascending=False
)

result['Peak_AQI'] = result['Peak_AQI'].round(2)
result['Peak_7Day_Avg'] = result['Peak_7Day_Avg'].round(2)

print(result)
```

```python
import pandas as pd

# Historical Diwali dates (Delhi)
diwali_dates = pd.to_datetime([
    '2015-11-11', '2016-10-30', '2017-10-19',
    '2018-11-07', '2019-10-27', '2020-11-14',
    '2021-11-04', '2022-10-24', '2023-11-12',
    '2024-10-31', '2025-10-20'
])

# Daily Delhi AQI
daily = (
    df_2026.loc[df_2026['city'].eq('Delhi')]
    .groupby('date')['aqi']
    .mean()
    .sort_index()
)

# Compare AQI before and after each Diwali
windows = {
    'Before_7_Days': (-7, -1),
    'Diwali_Day': (0, 0),
    'After_1_3_Days': (1, 3),
    'After_4_7_Days': (4, 7),
    'After_8_14_Days': (8, 14)
}

results = []

for year, festival in zip(range(2015, 2026), diwali_dates):
    row = {'Year': year, 'Diwali_Date': festival}

    for label, (start, end) in windows.items():
        values = daily.loc[
            festival + pd.Timedelta(days=start):
            festival + pd.Timedelta(days=end)
        ]

        expected_days = end - start + 1

        row[label] = (
            values.mean()
            if len(values) >= expected_days
            else float('nan')
        )

    results.append(row)

diwali_analysis = pd.DataFrame(results)

diwali_analysis['Change_After_1_3_Days'] = (
    diwali_analysis['After_1_3_Days']
    - diwali_analysis['Before_7_Days']
)

print(diwali_analysis.round(2).to_string(index=False))

# Overall comparison across years
print("\nAverage AQI around Diwali:")

print(
    diwali_analysis[windows.keys()]
    .mean()
    .round(2)
)
```

```python
# Compare post-Diwali AQI with pre-Diwali AQI
comparison = diwali_analysis[
    ['Year', 'Before_7_Days', 'After_1_3_Days']
].dropna().copy()

comparison['AQI_Change'] = (
    comparison['After_1_3_Days'] - comparison['Before_7_Days']
)

comparison['Percentage_Change'] = (
    comparison['AQI_Change'] / comparison['Before_7_Days'] * 100
)

print("Year-wise Diwali AQI changes:")
print(comparison.round(2).to_string(index=False))

print("\nSummary:")
print(comparison['AQI_Change'].describe().round(2))

print("\nYears with AQI increase:",
      (comparison['AQI_Change'] > 0).sum())

print("Years with AQI decrease:",
      (comparison['AQI_Change'] < 0).sum())

# Paired statistical test
from scipy.stats import wilcoxon

if (comparison['AQI_Change'] != 0).any():
    statistic, p_value = wilcoxon(
        comparison['After_1_3_Days'],
        comparison['Before_7_Days']
    )
    print("\nWilcoxon signed-rank test")
    print("Statistic:", round(statistic, 2))
    print("P-value:", round(p_value, 4))
```

```python
import pandas as pd
import numpy as np
from scipy.stats import wilcoxon, ttest_rel

results = []

for year, diwali_date in zip(range(2015, 2026), diwali_dates):

    diwali_date = pd.Timestamp(diwali_date)

    # Seven days before Diwali
    before_start = diwali_date - pd.Timedelta(days=7)
    before_end = diwali_date - pd.Timedelta(days=1)

    # Seven days after Diwali
    after_start = diwali_date + pd.Timedelta(days=1)
    after_end = diwali_date + pd.Timedelta(days=7)

    # Extract AQI observations from the Series
    before = daily.loc[before_start:before_end]
    after = daily.loc[after_start:after_end]

    before_mean = before.mean()
    after_mean = after.mean()

    # Require at least 5 valid observations
    if before.count() >= 5 and after.count() >= 5:

        change = after_mean - before_mean
        percentage_change = (change / before_mean) * 100

        results.append({
            "Year": year,
            "Before_AQI": round(before_mean, 2),
            "After_AQI": round(after_mean, 2),
            "AQI_Change": round(change, 2),
            "Percentage_Change": round(percentage_change, 2),
            "Before_Days": before.count(),
            "After_Days": after.count()
        })

# Create comparison table
comparison = pd.DataFrame(results).set_index("Year")

print("Seven-day AQI comparison:")
print(comparison)

# Overall statistics
print("\nOverall results:")

print("Mean AQI before:",
      round(comparison["Before_AQI"].mean(), 2))

print("Mean AQI after:",
      round(comparison["After_AQI"].mean(), 2))

print("Mean AQI change:",
      round(comparison["AQI_Change"].mean(), 2))

print("Median AQI change:",
      round(comparison["AQI_Change"].median(), 2))

print("Years AQI increased:",
      (comparison["AQI_Change"] > 0).sum())

print("Years AQI decreased:",
      (comparison["AQI_Change"] < 0).sum())

# Statistical tests
before_values = comparison["Before_AQI"]
after_values = comparison["After_AQI"]

wilcoxon_result = wilcoxon(after_values, before_values)
ttest_result = ttest_rel(after_values, before_values)

print("\nWilcoxon signed-rank test:")
print("Statistic:", wilcoxon_result.statistic)
print("P-value:", wilcoxon_result.pvalue)

print("\nPaired t-test:")
print("Statistic:", ttest_result.statistic)
print("P-value:", ttest_result.pvalue)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

# Obtain the regression design matrix
X = pd.DataFrame(
    dynamic_model.model.exog,
    index=dynamic_model.model.data.row_labels,
    columns=dynamic_model.model.exog_names
)

# Obtain the model residuals
residuals = pd.Series(
    dynamic_model.resid,
    index=X.index
)

# Calculate the observation-level score matrix
scores = X.mul(residuals, axis=0)

# Calculate the HAC covariance matrix
maxlags = 7

meat = scores.T @ scores

for lag in range(1, maxlags + 1):

    weight = 1 - lag / (maxlags + 1)

    # Match observations using actual calendar dates
    previous_scores = scores.reindex(
        scores.index - pd.Timedelta(days=lag)
    )

    previous_scores.index = scores.index

    valid = previous_scores.notna().all(axis=1)

    current = scores.loc[valid].to_numpy()
    previous = previous_scores.loc[valid].to_numpy()

    meat += weight * (
        current.T @ previous
        + previous.T @ current
    )

# Calculate the covariance matrix
X_array = X.to_numpy()

bread = np.linalg.inv(X_array.T @ X_array)

hac_covariance = bread @ meat @ bread

# Calculate standard errors and statistical tests
standard_errors = np.sqrt(np.diag(hac_covariance))

coefficients = dynamic_model.params.to_numpy()

z_values = coefficients / standard_errors

p_values = 2 * norm.sf(np.abs(z_values))

# Construct the results table
hac_results = pd.DataFrame({
    "Coefficient": coefficients,
    "HAC_Std_Error": standard_errors,
    "z_value": z_values,
    "p_value": p_values,
    "CI_Lower": coefficients - 1.96 * standard_errors,
    "CI_Upper": coefficients + 1.96 * standard_errors
}, index=dynamic_model.params.index)

print("Calendar-aware HAC regression results:\n")
print(hac_results.round(4))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

# Get the design matrix and residuals from the interaction model
X = year_interaction_model.model.exog
residuals = year_interaction_model.resid.to_numpy()
column_names = year_interaction_model.model.exog_names

# Calculate observation-level regression scores
scores = X * residuals[:, np.newaxis]

# Use actual calendar dates
dates = dynamic_data.index
date_positions = {date: i for i, date in enumerate(dates)}

# Calculate the HAC meat matrix
meat = scores.T @ scores

maxlags = 7

for lag in range(1, maxlags + 1):

    cross_products = np.zeros_like(meat)

    for i, date in enumerate(dates):

        previous_date = date - pd.Timedelta(days=lag)

        if previous_date in date_positions:

            j = date_positions[previous_date]

            cross_products += np.outer(
                scores[i],
                scores[j]
            )

    weight = 1 - lag / (maxlags + 1)

    meat += weight * (
        cross_products + cross_products.T
    )

# Calculate HAC covariance matrix
bread = np.linalg.pinv(X.T @ X)

hac_covariance = bread @ meat @ bread

# Calculate the fire coefficient for each year
yearly_hac_results = []

for year in sorted(dynamic_data["Year"].unique()):

    contrast = np.zeros(len(column_names))

    # Base fire coefficient (2015)
    contrast[column_names.index("Log_PH_Fire_Lag4")] = 1

    # Add the year-specific interaction coefficient
    if year != 2015:

        interaction_name = (
            f"Log_PH_Fire_Lag4:C(Year)[T.{year}]"
        )

        contrast[column_names.index(interaction_name)] = 1

    coefficient = contrast @ year_interaction_model.params.to_numpy()

    standard_error = np.sqrt(
        contrast @ hac_covariance @ contrast
    )

    z_value = coefficient / standard_error

    p_value = 2 * norm.sf(abs(z_value))

    lower_ci = coefficient - 1.96 * standard_error
    upper_ci = coefficient + 1.96 * standard_error

    yearly_hac_results.append({
        "Year": year,
        "Coefficient": coefficient,
        "HAC_SE": standard_error,
        "z_value": z_value,
        "p_value": p_value,
        "Lower_95_CI": lower_ci,
        "Upper_95_CI": upper_ci
    })

yearly_hac_results = pd.DataFrame(yearly_hac_results)

print(yearly_hac_results.round(4).to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

def calendar_hac_results(model, fire_variable, dates, maxlags=7):

    X = model.model.exog
    residuals = model.resid.to_numpy()

    dates = pd.DatetimeIndex(dates)

    scores = X * residuals[:, None]

    meat = scores.T @ scores

    date_lookup = pd.Series(
        np.arange(len(dates)),
        index=dates
    )

    for lag in range(1, maxlags + 1):

        previous_dates = dates - pd.Timedelta(days=lag)

        valid = previous_dates.isin(dates)

        current_positions = np.where(valid)[0]

        previous_positions = date_lookup.loc[
            previous_dates[valid]
        ].to_numpy()

        cross_product = (
            scores[current_positions].T
            @ scores[previous_positions]
        )

        weight = 1 - lag / (maxlags + 1)

        meat += weight * (
            cross_product + cross_product.T
        )

    bread = np.linalg.solve(
        X.T @ X,
        np.eye(X.shape[1])
    )

    hac_covariance = bread @ meat @ bread

    coefficient = model.params[fire_variable]

    variable_index = model.model.exog_names.index(
        fire_variable
    )

    standard_error = np.sqrt(
        hac_covariance[variable_index, variable_index]
    )

    z_value = coefficient / standard_error

    p_value = 2 * norm.sf(abs(z_value))

    lower_ci = coefficient - 1.96 * standard_error
    upper_ci = coefficient + 1.96 * standard_error

    return {
        "Coefficient": coefficient,
        "HAC_SE": standard_error,
        "Z_value": z_value,
        "P_value": p_value,
        "Lower_95_CI": lower_ci,
        "Upper_95_CI": upper_ci,
        "R_squared": model.rsquared
    }


# Calculate results for all four lag models
lag_hac_results = []

for lag, fitted_model in lag_results.items():

    fire_variable = f"Log_PH_Fire_Lag{lag}"

    result = calendar_hac_results(
        fitted_model,
        fire_variable,
        dynamic_data.index,
        maxlags=7
    )

    result["Fire_Lag"] = lag

    lag_hac_results.append(result)


# Display comparison
lag_hac_table = pd.DataFrame(lag_hac_results)

lag_hac_table = lag_hac_table.set_index("Fire_Lag")

print(lag_hac_table.round(4).to_string())
```

```python
# Get the primary dynamic model design matrix and residuals
X_dynamic = dynamic_model.model.exog
resid_dynamic = dynamic_model.resid.to_numpy()

# Get the actual dates
dates_dynamic = pd.DatetimeIndex(dynamic_data.index)

# Calculate observation-level scores
scores_dynamic = X_dynamic * resid_dynamic[:, None]

# Initialise HAC meat matrix
meat_dynamic = scores_dynamic.T @ scores_dynamic

# Create date lookup
date_lookup_dynamic = pd.Series(
    np.arange(len(dates_dynamic)),
    index=dates_dynamic
)

# Calculate calendar-aware HAC terms
maxlags = 7

for lag in range(1, maxlags + 1):

    previous_dates = dates_dynamic - pd.Timedelta(days=lag)

    valid = previous_dates.isin(dates_dynamic)

    current_positions = np.where(valid)[0]

    previous_positions = date_lookup_dynamic.loc[
        previous_dates[valid]
    ].to_numpy()

    cross_product = (
        scores_dynamic[current_positions].T
        @ scores_dynamic[previous_positions]
    )

    weight = 1 - lag / (maxlags + 1)

    meat_dynamic += weight * (
        cross_product + cross_product.T
    )

# Calculate bread matrix
bread_dynamic = np.linalg.solve(
    X_dynamic.T @ X_dynamic,
    np.eye(X_dynamic.shape[1])
)

# Calculate calendar-aware HAC covariance
hac_covariance_dynamic = (
    bread_dynamic @ meat_dynamic @ bread_dynamic
)

print("Covariance matrix shape:", hac_covariance_dynamic.shape)
print("Contains NaN:", np.isnan(hac_covariance_dynamic).any())
print("Contains infinity:", np.isinf(hac_covariance_dynamic).any())
```

```python
# Extract coefficients
coefficients = dynamic_model.params

# Calculate correct calendar-aware HAC standard errors
hac_standard_errors = pd.Series(
    np.sqrt(np.diag(hac_covariance_dynamic)),
    index=coefficients.index
)

# Calculate 95% confidence intervals
lower_ci = coefficients - 1.96 * hac_standard_errors
upper_ci = coefficients + 1.96 * hac_standard_errors

# Select main explanatory variables
selected_variables = [
    "Log_PH_Fire_Lag4",
    "AQI_Lag1",
    "T2M",
    "RH2M",
    "WS10M",
    "Wind_Sin",
    "Wind_Cos",
    "PRECTOTCORR"
]

# Create coefficient table
coef_data = pd.DataFrame({
    "Variable": selected_variables,
    "Coefficient": coefficients[selected_variables].values,
    "Lower_CI": lower_ci[selected_variables].values,
    "Upper_CI": upper_ci[selected_variables].values
})

# Plot coefficients and confidence intervals
plt.figure(figsize=(10, 6))

plt.errorbar(
    coef_data["Coefficient"],
    coef_data["Variable"],
    xerr=[
        coef_data["Coefficient"] - coef_data["Lower_CI"],
        coef_data["Upper_CI"] - coef_data["Coefficient"]
    ],
    fmt="o",
    capsize=4
)

plt.axvline(
    x=0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Dynamic Regression Coefficients with Calendar-Aware HAC 95% Confidence Intervals",
    fontsize=12
)

plt.xlabel("Estimated Coefficient")
plt.ylabel("Explanatory Variables")

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()
plt.show()
```

## 1. Agricultural Fire Activity

The analysis of VIIRS satellite fire detections across Punjab and Haryana from 2015 to 2025 revealed a strong seasonal concentration of vegetation fires.

- October and November accounted for approximately 600,000+ fire detections during the study period.
- May also showed substantial fire activity, indicating a second seasonal peak.
- October–November fire detections reached their highest annual total in 2016, with 100,737 detections.
- The annual total declined to 5,133 detections in 2025.

The findings demonstrate a strong seasonal pattern in vegetation-fire activity. However, satellite detections should not be interpreted as an exact count of agricultural burning events, and changes in satellite detection patterns must be considered when interpreting long-term declines.

## 2. Delhi AQI Patterns

The analysis of Delhi's daily Air Quality Index (AQI) from 2015 to 2025 revealed considerable seasonal and year-to-year variation in air pollution.

The October–November period was selected for closer investigation because it coincides with increased vegetation-fire activity in Punjab and Haryana.

The analysis also considered meteorological conditions, including temperature, relative humidity, wind speed, wind direction and precipitation, to investigate their relationships with Delhi's AQI.

It is important to recognise that AQI is influenced by multiple emission sources and atmospheric processes. Therefore, agricultural fires are investigated as one potential contributing factor rather than the sole explanation for Delhi's pollution.

## 3. Relationship Between Fire Activity and AQI

The correlation analysis identified a positive relationship between Punjab–Haryana fire detections and Delhi AQI.

The correlation increased across the examined lag periods:

- Lag 0: 0.4286
- Lag 2: 0.4466
- Lag 4: 0.4664
- Lag 6: To be reported from the lag analysis.

A four-day lag was selected for the primary dynamic regression model.

The final regression model included lagged fire activity, previous-day AQI, meteorological variables, and year and month controls.

The model produced an R-squared of 0.7797 using 663 observations.

The calendar-aware HAC results indicated a statistically significant positive association between four-day lagged fire activity and Delhi AQI (coefficient = 5.4047, p < 0.001).

This represents an association after accounting for the other included variables. It does not establish that agricultural fires directly caused the observed changes in Delhi AQI.

## 4. Influence of Meteorological Conditions

Meteorological conditions were included in the regression model to account for their relationship with Delhi's air pollution.

The calendar-aware HAC regression results showed:

| Variable | Coefficient | P-value |
|---|---:|---:|
| Temperature (T2M) | -2.4216 | 0.0041 |
| Relative Humidity (RH2M) | -1.1277 | 0.0006 |
| Wind Speed (WS10M) | -14.6414 | <0.001 |
| Precipitation | -0.9451 | 0.0871 |

Temperature, relative humidity and wind speed had statistically significant negative conditional associations with AQI in the model. Precipitation was not statistically significant at the 5% level.

Wind direction was represented using sine and cosine transformations because it is a circular variable.

These results describe conditional statistical relationships and should not be interpreted as proof of direct causal effects.

## 5. Diwali Analysis

The study examined changes in Delhi AQI around Diwali across the available years from 2015 to 2025.

The analysis compared AQI before and after Diwali using multiple event windows.

For the ten-day before-and-after comparison:

- Average AQI change: +64.01 points.
- Median AQI change: +90 points.
- AQI increased in 6 years and decreased in 3 years.
- Paired t-test p-value: 0.0764.
- Wilcoxon signed-rank test p-value: 0.0977.

Neither test reached the conventional 5% significance level.

The three-day comparison produced a statistically significant result, while the seven-day and ten-day comparisons did not.

These findings suggest that the observed AQI changes around Diwali depend on the selected event window. The results should therefore be interpreted cautiously rather than treated as definitive evidence of a Diwali effect.

## 6. Model Diagnostics and Limitations

Several diagnostic tests and sensitivity analyses were conducted to evaluate the reliability of the regression results.

- Multicollinearity: All calculated VIF values were below 5.
- Heteroskedasticity: The Breusch–Pagan test indicated evidence of heteroskedasticity.
- Residual normality: The Jarque–Bera and Shapiro–Wilk tests rejected the assumption of normally distributed residuals.
- Functional form: The RESET test indicated possible model misspecification.
- Sensitivity analysis: Removing the five most influential observations reduced the fire coefficient from 5.4047 to 4.9833, while the positive association remained statistically significant.

The model used calendar-aware HAC standard errors to account for temporal dependence while respecting the gaps between October–November observation periods.

Important limitations include:

1. VIIRS fire detections do not represent an exact count of agricultural burning events.
2. NASA POWER weather variables are gridded estimates rather than direct measurements from every location in Delhi.
3. Delhi's AQI monitoring network expanded substantially during the study period.
4. Other pollution sources and atmospheric processes may influence AQI.
5. The observational regression design does not establish causation.
